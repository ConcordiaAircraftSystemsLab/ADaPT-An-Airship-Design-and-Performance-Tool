# Underlying Tools

The Airship Design Tool combines several open-source computational tools. Each serves a distinct purpose.

## OpenMDAO

OpenMDAO forms the computational backbone of the multidisciplinary framework.

The airship disciplines are represented as components and connected through a common multidisciplinary model. The top-level `AirshipAnalysisGroup.py` assembles these components into the integrated airship analysis.

Conceptually:

$$ \boxed{\text{Airship physics}} \quad+\quad \boxed{\text{OpenMDAO coupling}} \quad\longrightarrow\quad \boxed{\text{Integrated airship model}}. $$

OpenMDAO therefore does not provide the airship equations themselves. Those equations are implemented by the Airship Design Tool. OpenMDAO provides the infrastructure through which the discipline models communicate and the coupled design problem is solved.

This distinction will be maintained throughout the Theory Guide.

## NumPy and SciPy

NumPy provides the numerical-array functionality required by the physical models, while SciPy provides additional scientific-computing functionality. Both are explicit dependencies of the project.

## pandas

pandas is primarily used in the analysis layer to organize, process, store, and interrogate datasets generated by design-space exploration and other studies.

For example, the design-space workflow produces CSV datasets that are subsequently used for trade-space analysis. The repository explicitly separates these studies from the reusable core model.

## Matplotlib

Matplotlib provides visualization capabilities for design-space, sensitivity, validation, payload-range, and optimization results.

## pymoo

pymoo provides evolutionary and multi-objective optimization capabilities used in selected optimization and validation workflows. It is included among the framework's explicit dependencies.

## Airship-Design Theory

The numerical software provides the computational infrastructure, but the physical models require an independent theoretical basis.

The principal airship reference currently identified for the framework is:

> Grant E. Carichner and Leland M. Nicolai, *Fundamentals of Aircraft and Airship Design, Volume 2—Airship Design and Case Studies*, AIAA Education Series, 2013.

The uploaded edition identifies Carichner and Nicolai as the authors and AIAA as the publisher.

Importantly, however, this reference will not automatically be assigned to every model in the framework.

The Theory Guide will distinguish between:

| Classification | Meaning |
|---|---|
| Fundamental | Derived directly from physical laws |
| Literature | Implemented directly from a cited source |
| Adapted | Literature formulation modified for this framework |
| Empirical | Correlation based on experimental/historical data |
| Derived | Obtained mathematically from other model quantities |
| Assumed | Engineering/modeling assumption |
| Technology assumption | Assumed present/future technology performance |
| User-defined | Supplied by the user |
| Calibrated | Determined from validation/calibration data |

This classification will become an important part of the Theory Guide.
